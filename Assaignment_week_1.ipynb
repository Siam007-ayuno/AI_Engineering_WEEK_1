### Q1 Simple calculator program

```python
a = input("Enter the 1st Number :")
b = input("Enter the 2nd Number :")
```

```python
a = float(a)                            #converting integer to float
b = float(b)
print("Converted Numbers :",a ,"," ,b)
```

```python
sum = a+b                            #Addition
print("Sum of the Numbers :",sum)
x = a-b                              #substraction
print("Substraction of a & b :", x)
```

```python
product = a*b                  # Multiplication using * arthimatic operation
print("Product of a and b is :", product)
```

```python
div = b/a                      #Doing division operation on a & b
print("Division of a & b :", div)
```

###Q3 Temperature Converter

```python
c = input("Give the temperature in Celsius :")
c = float(c) # converting input into float
print("Temperature in Celsius is :",c,"°C")
#calculation

F = (c*9/5)+32
print("Temperature in Farhenheit is :",F,"°F")
```
